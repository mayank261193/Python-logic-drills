```python
import pandas as pd
df = pd.read_csv("session6_policy_data.csv")
```

### Exercise 1 — Load and measure  ·  *easy*

The data is already loaded into `df` above. Print its shape (rows, columns).

```python
print(df.shape)     # (51, 5)
```

### Exercise 2 — First and last rows  ·  *easy*

Show the first 3 rows, then the last 2 rows.

```python
display(df.head(3))
display(df.tail(2))
```

### Exercise 3 — The column list  ·  *easy*

Print the column names.

```python
print(df.columns)   # Index(['region', 'year', 'employment_rate', 'budget_million', 'population'], ...)
```

### Exercise 4 — Spot the missing data  ·  *medium*

Use `.info()` to see non-null counts, then print how many `budget_million` values are present.

```python
df.info()
print(df["budget_million"].notna().sum())   # 49 present (2 missing)
```

### Exercise 5 — Summary statistics  ·  *medium*

Print `.describe()` and read off the largest population.

```python
print(df.describe())
# The max population is 67400
```

### Exercise 6 — A column is a Series  ·  *medium*

Pull out the `region` column and confirm its type is a Series.

```python
regions = df["region"]
print(type(regions))   # <class 'pandas.core.series.Series'>
```

### Exercise 7 — Select two columns  ·  *medium*

Show just `region` and `employment_rate` (double brackets → a DataFrame).

```python
df[["region", "employment_rate"]].head()
```

### Exercise 8 — Count the rows  ·  *medium*

Print the number of rows using `.shape` (not `len` of the file).

```python
print(df.shape[0])   # 51
```

### Exercise 9 — A quick average  ·  *hard*

Print the average `employment_rate` across the whole dataset, rounded to 2 dp.

```python
print(round(df["employment_rate"].mean(), 2))   # about 92.46
```

### Exercise 10 — Earliest and latest year  ·  *hard*

Print the minimum and maximum `year` in the data.

```python
print(df["year"].min(), df["year"].max())   # 2021 2030
```
